```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

```python
base = np.ones((128, 128, 3), dtype=np.uint8) * 255
```

```python
linspace = np.array([np.hstack((np.zeros(64).T, np.linspace(0, 64, 64)))]*128)
```

```python
grandient = (linspace > linspace.T)*linspace + (linspace <= linspace.T)*linspace.T
plt.imshow(grandient)
```

```python
plt.imshow(np.vstack((np.hstack((grandient[::2, ::2], grandient[::2, ::-2])), np.hstack((grandient[::-2, ::2], grandient[::-2, ::-2])))))
```

```python
base[:, :, 0] = grandient[::-1, ::-1]
base[:, :, 0] *= 1
base[:, :, 1] = np.vstack((np.hstack((grandient[::2, ::2], grandient[::2, ::-2])), np.hstack((grandient[::-2, ::2], grandient[::-2, ::-2]))))
base[:, :, 1] *= 0
base[:, :, 2] = grandient
base[:, :, 2] *= 1

plt.imshow(base)
```

```python
# base[:, :, 0] = np.vstack((np.hstack((grandient[::2, ::2], grandient[::2, ::-2])), np.hstack((grandient[::-2, ::2], grandient[::-2, ::-2]))))
base[:, :, 0] = grandient[::-1, ::-1]
base[:, :, 0] *= 1
base[:, :, 1] = grandient
base[:, :, 1] *= 1
base[:, :, 2] = grandient[::-1]
base[:, :, 2] *= 1

plt.imshow(base)
```

```python
plt.imsave('gradient.png', base)
```
